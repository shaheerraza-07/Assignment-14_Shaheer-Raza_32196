## 1. Import libraries and load the datasets


```python
import pandas as pd
import matplotlib.pyplot as plt

# Load both CSV files
ecommerce = pd.read_csv("pakistani_ecommerce_dataset.csv")
students = pd.read_csv("student_lifestyle_dataset.csv")

# Check the first five rows
print("E-commerce data:")
display(ecommerce.head())

print("Student lifestyle data:")
display(students.head())
```

# Dataset 1: Pakistani E-Commerce Data

## Q1. Which is the most popular product category?

I used `value_counts()` to count how many orders/records belong to each product category. The category with the highest count is the most popular in this dataset.


```python
category_counts = ecommerce["product_category"].value_counts()
display(category_counts)

most_popular_category = category_counts.idxmax()
print("Most popular product category:", most_popular_category)
```

## Q2. What is the average order value?

The average order value is calculated by taking the mean of the `order_value` column.


```python
average_order_value = ecommerce["order_value"].mean()
print("Average order value:", round(average_order_value, 2))
```

## Q3. Which city has the highest number of orders?

I counted the number of records for each city and compared the totals.


```python
orders_by_city = ecommerce["city"].value_counts()
display(orders_by_city)

city_with_most_orders = orders_by_city.idxmax()
print("City with the highest number of orders:", city_with_most_orders)
```

## Q4. What is the most used payment method?

I used the payment method column to count how many times each method appears.


```python
payment_counts = ecommerce["payment_method"].value_counts()
display(payment_counts)

most_used_payment = payment_counts.idxmax()
print("Most used payment method:", most_used_payment)
```

## Q5. Draw a bar chart showing average order value by city.

I grouped the data by city and calculated the average order value for each city. The bar chart makes the comparison easier to see.


```python
average_by_city = ecommerce.groupby("city")["order_value"].mean().sort_values(ascending=False)
display(average_by_city)

average_by_city.sort_values().plot(kind="bar", figsize=(7, 4))
plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average order value")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# Dataset 2: Student Lifestyle & Study Habits

## Q1. What is the average number of study hours?

I calculated the mean of the `study_hours` column to find the average study time.


```python
average_study_hours = students["study_hours"].mean()
print("Average study hours:", round(average_study_hours, 2))
```

## Q2. Who sleeps more on average – males or females?

I grouped the students by gender and calculated the average sleep hours for each group.


```python
average_sleep = students.groupby("gender")["sleep_hours"].mean()
display(average_sleep)

print("Gender with higher average sleep:", average_sleep.idxmax())
```

## Q3. Is there a relationship between screen time and sleep hours?

I calculated the Pearson correlation between screen time and sleep hours. A positive value means the two variables tend to increase together, while a negative value means one tends to decrease as the other increases. A value close to zero means there is little linear relationship.


```python
screen_sleep_corr = students["screen_time"].corr(students["sleep_hours"])
print("Correlation between screen time and sleep hours:", round(screen_sleep_corr, 3))
```

## Q4. Draw a histogram of daily transport cost.

The histogram groups daily transport costs into ranges and shows how many students fall within each range.


```python
plt.figure(figsize=(7, 4))
plt.hist(students["daily_transport_cost"].dropna(), bins=10, edgecolor="black")
plt.title("Daily Transport Cost")
plt.xlabel("Daily transport cost")
plt.ylabel("Number of students")
plt.tight_layout()
plt.show()
```

## Q5. Compare average internet usage by gender.

I grouped the data by gender and calculated the average internet usage for each group.


```python
average_internet = students.groupby("gender")["internet_usage"].mean()
display(average_internet)

print("Gender with higher average internet usage:", average_internet.idxmax())
```
